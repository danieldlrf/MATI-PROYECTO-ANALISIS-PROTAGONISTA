```python
import pandas as pd
import networkx as nx
import json
```

```python
# -----------------------------------------------------------------------------
# 1. JUEGO DE TRONOS (Procesar y Guardar)
# -----------------------------------------------------------------------------
df_got = pd.read_csv('../data/raw/got/got-s1-edges.csv')
df_got = df_got.rename(columns={'Source': 'source', 'Target': 'target', 'Weight': 'weight'})
df_got.to_csv('../data/processed/got/got_s1_edges_clean.csv', index=False, encoding='utf-8')

# -----------------------------------------------------------------------------
# 2. MARVEL (Procesar y Guardar)
# -----------------------------------------------------------------------------
df_marvel = pd.read_csv('../data/raw/marvel/marvel-unimodal-edges.csv')
df_marvel = df_marvel.rename(columns={'Source': 'source', 'Target': 'target', 'Weight': 'weight'})
df_marvel.to_csv('../data/processed/marvel/marvel_edges_clean.csv', index=False, encoding='utf-8')

# -----------------------------------------------------------------------------
# 3. STAR WARS (JSON -> CSV Procesado)
# -----------------------------------------------------------------------------
json_path = "../data/raw/starwars/starwars-episode-1-interactions-allCharacters.json"
csv_path = "../data/processed/starwars/starwars_ep1_edges_clean.csv"

with open(json_path, 'r', encoding='utf-8') as f:
    data = json.load(f)
    node_names = [node['name'] for node in data['nodes']]

    edges = []
    for link in data['links']:
        edges.append({
            'source': node_names[link['source']],
            'target': node_names[link['target']],
            'weight': link['value']
        })

    df_sw = pd.DataFrame(edges)
    df_sw.to_csv(csv_path, index=False, encoding='utf-8')
```
