```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from pyvis.network import Network

# 1. Cargar los CSV globales estandarizados
df_got = pd.read_csv('../data/processed/got/got_full_interactions_edges_clean.csv')
df_sw = pd.read_csv('../data/processed/starwars/starwars_full_interactions_edges_clean.csv')
df_marvel = pd.read_csv('../data/processed/marvel/marvel_edges_clean.csv') 

# 2. Convertir cada DataFrame en un objeto Grafo de NetworkX
G_got = nx.from_pandas_edgelist(df_got, source='source', target='target', edge_attr='weight', create_using=nx.Graph())
G_sw = nx.from_pandas_edgelist(df_sw, source='source', target='target', edge_attr='weight', create_using=nx.Graph())
G_marvel = nx.from_pandas_edgelist(df_marvel, source='source', target='target', edge_attr='weight', create_using=nx.Graph())


net = Network(height="750px", width="100%", bgcolor="#222222", font_color="white", notebook=False)
net.from_nx(G_got)

# Activa físicas interactivas y guarda el HTML
net.show_buttons(filter_=['physics'])
net.write_html("got_interactivo.html")
```
