```python
import pandas as pd
import json
import glob
import os
```

```python
# -----------------------------------------------------------------------------
# 1. JUEGO DE TRONOS 
# -----------------------------------------------------------------------------

file = "../data/processed/got/got_full_interactions_edges_clean.csv"

input_files = [f for f in sorted(glob.glob("../data/processed/got/*.csv")) if os.path.basename(f) != "got_full_interactions_edges_clean.csv"]

df_match = pd.concat([pd.read_csv(f)[['source', 'target', 'weight']] for f in input_files], ignore_index=True)

pair = df_match[['source', 'target']].apply(sorted, axis=1)
df_match['source'] = [p[0] for p in pair]
df_match['target'] = [p[1] for p in pair]

df_got_global = df_match.groupby(['source', 'target'], as_index=False)['weight'].sum()
df_got_global.to_csv(file, index=False, encoding='utf-8')


# -----------------------------------------------------------------------------
# 2. STAR WARS
# -----------------------------------------------------------------------------
json_path = "../data/raw/starwars/starwars-full-interactions-allCharacters-merged.json"
csv_path = "../data/processed/starwars/starwars_full_interactions_edges_clean.csv"

with open(json_path, 'r', encoding='utf-8') as f:
    data = json.load(f)
    node_names = [node['name'] for node in data['nodes']]

    edges = []
    for link in data['links']:
        edges.append({
            'source': node_names[link['source']],
            'target': node_names[link['target']],
            'weight': link['value']
        })

    df_sw = pd.DataFrame(edges)
    df_sw.to_csv(csv_path, index=False, encoding='utf-8')
```
